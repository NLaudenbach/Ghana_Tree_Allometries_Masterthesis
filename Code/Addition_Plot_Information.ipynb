## **XLSX-Paths**

```python
# ============================================================
# PLOT-DATEIEN
# ============================================================

plot_files = {

    "KO4": r"Z:\Ghana\RCT_outputs\KO4_preseg_RCT\output\rct\segmented\KO4_Trees_Inventory\KO4_tree_inventory_master.xlsx",

    # Weitere Plots einfach hier ergänzen:
    #
    # "KO2": r"Z:\...\KO2_tree_inventory_master.xlsx",
    # "KO5": r"Z:\...\KO5_tree_inventory_master.xlsx",
    # "BO1": r"Z:\...\BO1_tree_inventory_master.xlsx",
    # "BO2": r"Z:\...\BO2_tree_inventory_master.xlsx",
    # "ANK": r"Z:\...\ANK_tree_inventory_master.xlsx",
}
```

## **Addition of Information**

```python
# ============================================================
# ZUORDNUNG DER PLOTS ZU DEN STANDORTEN AUS DEM PAPER
# ============================================================

plot_to_site = {

    # Ankasa
    "ANK": "Ankasa",

    # Bobiri
    "BO1": "Bobiri",
    "BO2": "Bobiri",

    # Kogyae
    "KO2": "Kogyae",
    "KO4": "Kogyae",
    "KO5": "Kogyae",
}

# ============================================================
# KLIMA- UND STANDORTDATEN AUS DER TABELLE
# ============================================================

site_data = {

    "Ankasa": {
        "Mean_annual_air_temperature_TH": 25.0,
        "Mean_annual_precipitation_TH": 2050,
        "Mean_maximum_climatological_water_deficit_TH": -13,
        "Soil_pH_TH": 4.27,
        "Leaf_residence_time_TH": "9-10",
    },

    "Bobiri": {
        "Mean_annual_air_temperature_TH": 25.7,
        "Mean_annual_precipitation_TH": 1500,
        "Mean_maximum_climatological_water_deficit_TH": -374,
        "Soil_pH_TH": 6.05,
        "Leaf_residence_time_TH": "4-6",
    },

    "Kogyae": {
        "Mean_annual_air_temperature_TH": 26.4,
        "Mean_annual_precipitation_TH": 1200,
        "Mean_maximum_climatological_water_deficit_TH": -412,
        "Soil_pH_TH": 6.07,
        "Leaf_residence_time_TH": "6.5-8",
    },
}

# ============================================================
# FUNKTION: KLIMADATEN AN EINE BAUMTABELLE ANHÄNGEN
# ============================================================

def add_site_data(plot_name, input_file):

    print("\n" + "=" * 60)
    print(f"Verarbeite Plot: {plot_name}")
    print(f"Datei: {input_file}")

    # --------------------------------------------------------
    # Prüfen, ob Datei existiert
    # --------------------------------------------------------

    if not os.path.exists(input_file):
        print("FEHLER: Datei wurde nicht gefunden.")
        return

    # --------------------------------------------------------
    # Prüfen, ob Plot eine Standort-Zuordnung besitzt
    # --------------------------------------------------------

    if plot_name not in plot_to_site:
        print(f"FEHLER: Für {plot_name} wurde kein Standort definiert.")
        return

    site = plot_to_site[plot_name]

    print(f"Zugeordneter Standort: {site}")

    # --------------------------------------------------------
    # Excel einlesen
    # --------------------------------------------------------

    df = pd.read_excel(input_file)

    print(f"Bäume in Tabelle: {len(df)}")
    print(f"Bestehende Spalten: {len(df.columns)}")

    # --------------------------------------------------------
    # Standortdaten hinzufügen
    # --------------------------------------------------------

    for column_name, value in site_data[site].items():
        df[column_name] = value

    # --------------------------------------------------------
    # Ausgabe-Datei erzeugen
    # --------------------------------------------------------

    input_dir = os.path.dirname(input_file)
    input_filename = os.path.basename(input_file)

    filename_without_extension = os.path.splitext(input_filename)[0]

    output_file = os.path.join(
        input_dir,
        filename_without_extension + "_TH.xlsx"
    )

    # --------------------------------------------------------
    # Speichern
    # --------------------------------------------------------

    df.to_excel(
        output_file,
        index=False
    )

    print(f"Neue Spalten hinzugefügt: {len(site_data[site])}")
    print(f"Neue Gesamtzahl Spalten: {len(df.columns)}")

    print("Gespeichert als:")
    print(output_file)

# ============================================================
# ALLE PLOTS VERARBEITEN
# ============================================================

for plot_name, input_file in plot_files.items():

    add_site_data(
        plot_name=plot_name,
        input_file=input_file
    )


print("\n" + "=" * 60)
print("FERTIG")
print("=" * 60)
```
